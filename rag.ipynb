```python
import os
from dotenv import load_dotenv

from langchain_community.document_loaders import TextLoader

from langchain_text_splitters import RecursiveCharacterTextSplitter

load_dotenv()
```

```python
DATA_FILE_PATH = os.path.join('data', 'hr_policy.txt')
print("Data file path:", DATA_FILE_PATH)
```

```python
loader = TextLoader(DATA_FILE_PATH, encoding='utf-8')

documents = loader.load()

print("Documents loaded successfully.")

print("=" * 40 )
print(documents)


#Langchain processses everything in the form of documents. 
# So we need to convert our text into documents. 
# We can use the TextLoader class from langchain_community.document_loaders to load the text file 
# and convert it into documents.

#Page_content is the actual content of the document.
#Metadata is the information about the document. It can be anything like the source of the document, 
#the author of the document, etc.

len(documents)
#Only 1 document as of now. We can keep as many documents as we want.

documents[0].page_content

documents[0].metadata
```

```python
print(f"Total no of characters in the document: {len(documents[0].page_content)}")
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50,
)

chunks = text_splitter.split_documents(documents)

print(chunks)

print(f"Total no of chunks created: {len(chunks)}")

#Now each chunk is a document in itself. We can see that each chunk has a page_content and metadata.

```

```python
print(chunks[0])

print("-" * 40)

print(chunks[7].page_content)
```

```python
from langchain_community.embeddings import JinaEmbeddings

embeddings_model = JinaEmbeddings(
    model_name="jina-embeddings-v2-base-en"
)

print("Embeddings model loaded successfully -> ", embeddings_model.model_name)

print(embeddings_model)
```

```python
#Just checking how the vector looks like for a sample text.

text = "Employees get 12 sick leaves per year."

vector = embeddings_model.embed_query(text)

print("Number of dimensions:", len(vector))
print("Vector:")
print(vector)
```

```python
#Store the embeddings in a vector database. 
# We can use any vector database like Pinecone, Weaviate, Milvus, etc.

from langchain_community.vectorstores import FAISS

vectorstore = FAISS.from_documents(chunks, embeddings_model)

print("Chunks stored !!!", vectorstore.index.ntotal)
```

```python
#Similarity search -- FAISS

test_query = "How many sick leaves employees get?"
top_matches = vectorstore.similarity_search(test_query, k=2)
print(f"Query: {test_query}")
for i, match in enumerate(top_matches, start=1):
    print(f"Match {i} : ")
    print(match.page_content)
    print("-" * 40) 
```

```python
#TOOLS

retriever = vectorstore.as_retriever(search_kwargs={"k": 3})

def search_hr_policy(question: str)-> str:
    """Search HR policy documents for a specific question related to leave, work from home, 
    probation, notice period, reimbursement, code of conduct, holidays, or exit process."""
    matching_chunks = retriever.invoke(question)

    return "\n\n".join([chunk.page_content for chunk in matching_chunks])
```

```python
#DATA Retrival

from langchain_openrouter import ChatOpenRouter

llm = ChatOpenRouter(
    model="qwen/qwen3.8-27b",
    #api_key=openrouter_api_key,
    #tools=["jina-embeddings-v2-base-en"],
    temperature=0.7,
    max_tokens=1000
)

#llm.model_name

test_response = llm.invoke("Is learning RAG diffficult?")

test_response.content
```

```python
#AI AGENT
from langchain.agents import create_agent

hr_assistant = create_agent(
    model=llm,
    tools=[search_hr_policy],
    system_prompt="""
    
    You are a helpful HR assistant. You have access to the HR policy document and have to answer questions related to it.
    If the answer is not in the document, you have to say "I don't know". You should not make up answers.

    """
)
print("HR Assistant created successfully.")
```

```python
def ask_hr_assistant(question: str) -> str:
    """
    Ask a question to the HR assistant and get the nicely formatted answer."""

    print("=" * 40)
    print(f"Question: {question}")
    print("-" * 40)


    response = hr_assistant.invoke({
        "messages": [
            {"role": "user", "content": question},
        ]
    })
    answer = response["messages"][-1].content
    print(f"Answer: {answer}")
    print("=" * 40)

    print()
    return answer

    
```
